## 核心知识

Figure 是整张图，Axes 是其中的一块绘图区。`plt.figure(figsize=(w,h), dpi=...)` 的尺寸单位是英寸。通过 `fig, ax = plt.subplots()` 也能同时创建 Figure 与 Axes。位图像素大小约为英寸尺寸乘以导出 dpi；每次绘图最好显式持有 Figure 对象。

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
OUTPUT = Path.cwd() / "matplotlib_练习输出"
OUTPUT.mkdir(exist_ok=True)
x = np.linspace(0, 2 * np.pi, 150)
```

## 示例：观察画布尺寸

下面创建 6×3 英寸画布，并查看对象自身记录的尺寸。

```python
fig, ax = plt.subplots(figsize=(6, 3), dpi=100)
ax.plot(x, np.sin(x))
ax.set_title("6 × 3 inch figure")
print("Figure inches:", fig.get_size_inches(), "dpi:", fig.dpi)
plt.show()
```

## 示例：同一 Figure 的两张图

Axes 是 Figure 的子图，可分别控制标题与坐标范围。

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 3), constrained_layout=True)
axes[0].plot(x, np.sin(x)); axes[0].set_title("sin")
axes[1].plot(x, np.cos(x), color="tomato"); axes[1].set_title("cos")
fig.suptitle("One Figure, two Axes")
plt.show()
```
