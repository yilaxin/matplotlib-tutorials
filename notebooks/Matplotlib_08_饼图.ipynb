# Matplotlib 08：饼图

`pie()` 用扇区面积表达各类别占比。`autopct` 显示百分比，`startangle` 旋转起始角，`explode` 突出指定扇区。

```python
import matplotlib.pyplot as plt
labels=['学习','工作','运动','休息']; values=[35,30,15,20]
plt.pie(values,labels=labels,autopct='%1.1f%%',startangle=90,explode=[.06,0,0,0],shadow=True)
plt.title('一天时间分配'); plt.axis('equal'); plt.show()
```

## 环形图

设置 `wedgeprops={'width':...}` 可以把饼图变成环形图，更适合展示百分比构成。

```python
labels=['完成','进行中','未开始']; values=[60,25,15]
plt.pie(values,labels=labels,autopct='%1.0f%%',wedgeprops={'width':.38,'edgecolor':'white'},startangle=90)
plt.title('项目进度'); plt.axis('equal'); plt.show()
```

注意：类别过多或数值差异很小时，柱形图通常比饼图更容易比较。练习：修改颜色并突出占比最大的部分。
