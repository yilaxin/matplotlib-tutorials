# Matplotlib 06：散点图

`scatter()` 适合观察两个变量的关系。点的大小由 `s` 控制，颜色由 `c` 和 `cmap` 控制，透明度由 `alpha` 控制。

```python
import matplotlib.pyplot as plt
import numpy as np
rng=np.random.default_rng(10); x=rng.normal(50,10,80); y=.7*x+rng.normal(0,7,80)
size=rng.uniform(20,180,80); color=y
plt.scatter(x,y,s=size,c=color,cmap='viridis',alpha=.75,edgecolors='white')
plt.colorbar(label='Y 值'); plt.xlabel('广告投入'); plt.ylabel('销售额'); plt.title('投入与销售额的散点关系'); plt.show()
```

## 分组散点图

按类别分别调用 `scatter()`，可以使用不同颜色和图例表达分组信息。

```python
a=rng.normal(0,1,(30,2)); b=rng.normal(3,1,(30,2))
plt.scatter(a[:,0],a[:,1],label='组 A',alpha=.7); plt.scatter(b[:,0],b[:,1],label='组 B',alpha=.7)
plt.axhline(0,color='gray',linewidth=.7); plt.legend(); plt.title('两组样本'); plt.show()
```
