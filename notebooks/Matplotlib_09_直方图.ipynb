# Matplotlib 09：直方图

直方图把连续数据划分成多个区间（bins），统计每个区间的频数，用于观察数据分布、偏态和离群值。

```python
import matplotlib.pyplot as plt
import numpy as np
rng=np.random.default_rng(3); scores=rng.normal(75,10,1000)
plt.hist(scores,bins=20,color='mediumpurple',edgecolor='white',alpha=.85)
plt.axvline(scores.mean(),color='red',linestyle='--',label=f'平均值={scores.mean():.1f}')
plt.xlabel('分数'); plt.ylabel('人数'); plt.title('考试成绩分布'); plt.legend(); plt.show()
```

## 比较两组分布

将两组数据叠加时使用 `alpha`，可以直观看到区间重叠。`density=True` 会把纵轴转换为概率密度。

```python
a=rng.normal(70,8,600); b=rng.normal(78,8,600)
plt.hist(a,bins=18,alpha=.55,density=True,label='班级 A'); plt.hist(b,bins=18,alpha=.55,density=True,label='班级 B')
plt.xlabel('分数'); plt.ylabel('密度'); plt.title('两组成绩分布比较'); plt.legend(); plt.show()
```

练习：分别使用 5、10、30 个 bins，比较分组数量对图形解读的影响。
