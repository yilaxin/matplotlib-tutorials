# Matplotlib 04：网格线

`grid()` 能帮助读者从数据点对应到坐标轴。可通过 `axis`、`which`、`linestyle`、`alpha` 细化网格。

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,2*np.pi,200)
plt.plot(x,np.sin(x),label='sin(x)'); plt.plot(x,np.cos(x),label='cos(x)')
plt.grid(True,which='major',axis='both',linestyle='--',color='gray',alpha=.45)
plt.xticks([0,np.pi/2,np.pi,3*np.pi/2,2*np.pi],['0','π/2','π','3π/2','2π']); plt.legend(); plt.show()
```

## 只显示单方向网格

当重点是比较横向或纵向数值时，可以使用 `axis='y'` 或 `axis='x'`，让图面更加简洁。

```python
values=[32,45,28,51,40]; names=['A','B','C','D','E']
plt.bar(names,values,color='skyblue'); plt.grid(axis='y',alpha=.4); plt.ylabel('数量'); plt.title('只显示水平网格'); plt.show()
```
