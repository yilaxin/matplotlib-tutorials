# Matplotlib 01：绘图标记

本节学习 `marker`、`markersize`、`markeredgecolor`、`markerfacecolor`，为折线图的数据点增加清晰的视觉提示。

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.arange(1,7); y=np.array([12,18,15,22,19,25])
plt.plot(x,y,marker='o',markersize=9,markerfacecolor='orange',markeredgecolor='navy',linewidth=2)
plt.xticks(x); plt.xlabel('月份'); plt.ylabel('销量'); plt.title('带标记的销量趋势'); plt.grid(alpha=.3); plt.show()
```

## 常见标记比较

`o` 为圆形、`s` 为方形、`^` 为上三角、`*` 为星号。可以用 `markevery` 只标记部分数据点，避免数据量较大时过于拥挤。

```python
markers=['o','s','^','*']; names=['圆形','方形','三角形','星号']
for i,(m,n) in enumerate(zip(markers,names)):
    plt.plot(x, y+i*2, marker=m, label=n, markersize=7)
plt.legend(); plt.title('不同 marker 的效果'); plt.show()
```

## 小结与练习

练习：将上面的折线改为红色虚线，并只显示每隔两个点的标记（提示：`markevery=2`）。
