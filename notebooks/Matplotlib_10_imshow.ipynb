# Matplotlib 10：`imshow()`

`imshow()` 将二维数组映射为图像颜色，适合矩阵、热力图和像素数据。`cmap` 控制颜色方案，`aspect` 控制纵横比。

```python
import matplotlib.pyplot as plt
import numpy as np
matrix=np.arange(100).reshape(10,10)
plt.imshow(matrix,cmap='viridis',aspect='auto'); plt.colorbar(label='数值')
plt.title('二维数组热力图'); plt.xlabel('列'); plt.ylabel('行'); plt.show()
```

## 用数学函数生成热力图

下面生成一个二维高斯分布。`origin='lower'` 让原点位于左下角；`extent` 可以设置坐标轴的实际范围。

```python
x=np.linspace(-3,3,200); y=np.linspace(-3,3,160); X,Y=np.meshgrid(x,y)
Z=np.exp(-(X**2+Y**2)/2)
plt.imshow(Z,extent=[-3,3,-3,3],origin='lower',cmap='magma'); plt.colorbar(label='概率密度')
plt.contour(X,Y,Z,colors='white',levels=6,linewidths=.6); plt.xlabel('X'); plt.ylabel('Y'); plt.title('二维高斯热力图'); plt.show()
```
