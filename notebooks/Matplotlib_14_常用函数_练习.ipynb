## 核心知识

`pyplot` 会操作当前图表；Axes 接口明确指出目标图，例如 `ax.plot()`、`ax.set_title()`。多子图练习推荐使用 `fig, axes = plt.subplots(...)`。常用函数按任务分为：创建图、绘制数据、标注/坐标、布局、导出。需要时用 `help(ax.plot)` 或官方文档查具体参数。

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(14)
x = np.arange(1, 7)
sales = np.array([12, 15, 14, 18, 20, 23])
```

## 示例：同一数据的两种调用方式

`plt.plot()` 和 `ax.plot()` 都可以作图。后者在多子图中更容易区分目标。

```python
fig, ax = plt.subplots()
ax.plot(x, sales, marker="o", label="sales")
ax.set(title="Sales trend", xlabel="Month", ylabel="Units")
ax.legend()
plt.show()
```

## 示例：一个 Figure 上的两个 Axes

两个子图可分别使用不同的图表类型，而不相互覆盖。

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x, sales, marker="o")
axes[0].set_title("Trend")
axes[1].bar(x, sales, color="teal")
axes[1].set_title("Comparison")
fig.tight_layout()
plt.show()
```
