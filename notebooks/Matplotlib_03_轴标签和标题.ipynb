# Matplotlib 03：轴标签和标题

使用 `xlabel()`、`ylabel()`、`title()` 说明图表含义；`fontsize`、`color`、`loc` 可以调整标题样式和位置。

```python
import matplotlib.pyplot as plt
import numpy as np
days=np.arange(1,8); temp=[18,20,21,19,23,25,24]
plt.plot(days,temp,marker='o')
plt.xlabel('日期（天）',fontsize=12); plt.ylabel('温度（℃）',fontsize=12)
plt.title('一周温度变化',fontsize=16,color='navy',pad=15); plt.xticks(days); plt.show()
```

## 中文显示与注释

如果系统没有合适的中文字体，可将 `font.sans-serif` 改为本机已有字体。`annotate()` 可以为关键点添加说明。

```python
plt.rcParams['axes.unicode_minus']=False
plt.plot(days,temp,marker='o'); peak=int(np.argmax(temp))
plt.annotate(f'最高：{temp[peak]}℃',(days[peak],temp[peak]),xytext=(days[peak]-1,temp[peak]+2),arrowprops={'arrowstyle':'->'})
plt.xlabel('日期'); plt.ylabel('温度'); plt.title('带注释的折线图'); plt.show()
```

练习：将标题改为左对齐或右对齐，并为最低温度点添加注释。
