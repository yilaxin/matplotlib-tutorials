# Matplotlib 05：绘制多图

`subplots()` 返回 Figure 和 Axes，可以在一个画布中安排多个子图；`sharex`、`sharey` 可共享坐标轴。

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,2*np.pi,200)
fig,axes=plt.subplots(2,2,figsize=(10,7),sharex=True)
funcs=[(np.sin,'sin'),(np.cos,'cos'),(lambda x:np.sin(2*x),'sin(2x)'),(lambda x:np.cos(2*x),'cos(2x)')]
for ax,(fn,name) in zip(axes.flat,funcs):
    ax.plot(x,fn(x)); ax.set_title(name); ax.grid(alpha=.25)
fig.suptitle('四个函数子图',fontsize=16); fig.tight_layout(); plt.show()
```

## 使用 `subplot()`

`plt.subplot(1,2,1)` 表示 1 行 2 列中的第 1 个区域，适合快速创建少量子图。

```python
plt.figure(figsize=(9,3))
plt.subplot(1,2,1); plt.plot(x,np.sin(x)); plt.title('左图')
plt.subplot(1,2,2); plt.plot(x,np.cos(x),color='orange'); plt.title('右图')
plt.tight_layout(); plt.show()
```
