## 核心知识

设置了 `label` 的绘图对象可自动加入图例。`loc` 控制图例相对位置，`bbox_to_anchor` 用于精确锚定，`ncol` 决定列数。若图例位于绘图区之外，配合 `constrained_layout` 或 `bbox_inches='tight'` 可减少裁切。

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2 * np.pi, 160)
```

## 示例：三条曲线的图例

每条曲线设置不同的 `label`，再统一调用 `ax.legend()`。

```python
fig, ax = plt.subplots(figsize=(7, 4))
for k in (1, 2, 3):
    ax.plot(x, np.sin(k * x), label=f"sin({k}x)")
ax.legend(loc="upper right", title="Functions", frameon=True)
ax.set_title("Multiple curves")
plt.show()
```

## 示例：图例放到右侧

右侧留出位置，可减少线条与图例重叠。

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(x, np.sin(x), label="sine")
ax.plot(x, np.cos(x), label="cosine")
ax.legend(loc="upper left", bbox_to_anchor=(1.02, 1), borderaxespad=0)
ax.set_title("Legend outside axes")
fig.tight_layout()
plt.show()
```
