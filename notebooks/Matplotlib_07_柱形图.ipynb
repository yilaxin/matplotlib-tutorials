# Matplotlib 07：柱形图

`bar()` 用于垂直柱形图，`barh()` 用于水平柱形图。柱形图适合比较离散类别的数量或大小。

```python
import matplotlib.pyplot as plt
import numpy as np
names=['一季度','二季度','三季度','四季度']; a=[120,145,132,170]; b=[105,150,140,160]
x=np.arange(len(names)); w=.35
plt.bar(x-w/2,a,w,label='产品 A'); plt.bar(x+w/2,b,w,label='产品 B')
plt.xticks(x,names); plt.ylabel('销量'); plt.title('两种产品季度销量'); plt.legend(); plt.grid(axis='y',alpha=.3); plt.show()
```

## 水平柱形图与数值标签

类别名称较长时，`barh()` 更易阅读。可以在柱子末端写出具体数值。

```python
items=['数据分析课程','Python 编程课程','可视化课程','机器学习课程']; values=[86,72,94,65]
bars=plt.barh(items,values,color='teal'); plt.xlim(0,100); plt.xlabel('满意度（%）'); plt.title('课程满意度')
for bar,v in zip(bars,values): plt.text(v+1,bar.get_y()+bar.get_height()/2,str(v),va='center')
plt.show()
```
