# Matplotlib 02：绘图线

本节学习颜色 `color`、线型 `linestyle`、线宽 `linewidth` 和透明度 `alpha`，并比较不同线型。

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,10,200)
styles=[('-', '实线'),('--','破折线'),('-.','点划线'),(':','虚线')]
for i,(style,label) in enumerate(styles):
    plt.plot(x,np.sin(x)+i*.7,linestyle=style,linewidth=2,label=label)
plt.legend(); plt.title('四种常见线型'); plt.grid(alpha=.25); plt.show()
```

## 多条线与透明度

绘制多组数据时应使用图例区分含义；`alpha` 可降低重叠曲线的不透明度。

```python
for phase,color in zip([0,.5,1],['tab:blue','tab:orange','tab:green']):
    plt.plot(x,np.sin(x+phase),color=color,alpha=.65,linewidth=3,label=f'phase={phase}')
plt.xlabel('x'); plt.ylabel('sin(x+phase)'); plt.legend(); plt.show()
```
