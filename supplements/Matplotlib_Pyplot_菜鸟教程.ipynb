# 第 1 课：Matplotlib Pyplot

Pyplot 是 Matplotlib 的子库，提供了类似 MATLAB 的绘图 API，可以方便地绘制二维图表。本课从菜鸟教程的 [Matplotlib Pyplot](https://www.runoob.com/matplotlib/matplotlib-pyplot.html) 一节开始，学习 `plot()` 的基本用法。

本课重点：
- 导入 `matplotlib.pyplot` 并使用别名 `plt`
- 使用 `plot()` 绘制折线和点
- 理解颜色、线型和标记的格式字符串
- 绘制多条曲线

```python
import matplotlib.pyplot as plt
import numpy as np

print('NumPy 版本：', np.__version__)
print('Matplotlib 已成功导入')
```

## 1. 绘制一条最简单的直线

`plot(x, y)` 会把 x、y 中对应的坐标连接起来。下面通过两个坐标 `(0, 0)` 和 `(6, 100)` 绘制一条直线。

`show()` 用于显示当前图像；在 Jupyter Notebook 中也可以省略，但保留它有助于理解完整的 Matplotlib 程序。

```python
xpoints = np.array([0, 6])
ypoints = np.array([0, 100])

plt.plot(xpoints, ypoints)
plt.show()
```

## 2. 绘制任意两个坐标点之间的线

把坐标 `(1, 3)` 和 `(8, 10)` 分别放入 x、y 数组即可。注意：x 和 y 中的元素数量必须相同。

```python
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

plt.plot(xpoints, ypoints)
plt.show()
```

## 3. 只绘制点：使用标记参数

如果只想显示坐标点而不绘制连接线，可以在 `plot()` 中传入格式字符串 `'o'`。其中 `o` 表示实心圆标记。

```python
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

plt.plot(xpoints, ypoints, 'o')
plt.show()
```

## 4. 绘制多个点和不规则折线

只要 x、y 数组长度一致，就可以绘制任意数量的点。下面的四个坐标会按照数组顺序连接，形成一条不规则折线。

```python
xpoints = np.array([1, 2, 6, 8])
ypoints = np.array([3, 8, 1, 10])

plt.plot(xpoints, ypoints, marker='o')
plt.show()
```

## 5. 省略 x 轴数据

当只传入 `y` 时，Matplotlib 会自动把 x 设置为 `0, 1, 2, ..., N-1`。这适合快速查看一组按顺序排列的数据。

```python
ypoints = np.array([3, 8, 1, 10, 5, 7])

plt.plot(ypoints, marker='o')
plt.xlabel('自动生成的 x（索引）')
plt.ylabel('y')
plt.show()
```

## 6. 颜色、线型和标记

`plot()` 的格式字符串可以同时指定颜色、标记和线型。例如：`'r--o'` 表示红色、虚线、圆形标记。也可以使用关键字参数分别设置 `color`、`linestyle`、`marker` 和 `linewidth`。

常见颜色：`b` 蓝色、`g` 绿色、`r` 红色、`k` 黑色；常见线型：`-` 实线、`--` 破折线、`-.` 点划线、`:` 虚线。

```python
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 1, 3, 2])

plt.plot(x, y, 'r--o', linewidth=2, markersize=8)
plt.title('格式字符串示例')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, alpha=0.3)
plt.show()
```

## 7. 在同一张图中绘制正弦和余弦曲线

一次 `plot()` 调用中可以传入多组 x、y 数据。下面使用 `np.arange()` 创建从 `0` 到 `4π` 的数据，并同时绘制正弦曲线和余弦曲线。

```python
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)
z = np.cos(x)

plt.plot(x, y, label='sin(x)')
plt.plot(x, z, label='cos(x)')
plt.title('正弦和余弦曲线')
plt.xlabel('x')
plt.ylabel('函数值')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 小结

- `plt.plot(x, y)` 是绘制二维折线图的基本方法。
- 只传入 `y` 时，x 会自动使用数组索引。
- `'o'`、`'r--'` 等格式字符串可以快速设置标记、颜色和线型。
- `label` 配合 `plt.legend()` 可以添加图例。
