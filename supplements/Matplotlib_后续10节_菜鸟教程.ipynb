# Matplotlib 后续 10 节：常用图表

本 Notebook 接续《Matplotlib Pyplot》，整理菜鸟教程后续 10 节内容。每节都包含简要说明和可以直接运行的示例。

参考目录：<https://www.runoob.com/matplotlib/matplotlib-tutorial.html>

```python
import matplotlib.pyplot as plt
import numpy as np

print('Matplotlib 和 NumPy 已导入')
```

## 1. Matplotlib 绘图标记

使用 `marker` 为数据点添加标记；常用标记包括 `o` 圆、`s` 方形、`^` 三角形和 `*` 星号。`markersize` 和 `markerfacecolor` 可以调整大小和填充色。

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 5, 3, 6, 4])

plt.plot(x, y, marker='o', markersize=10, markerfacecolor='orange',
         markeredgecolor='navy', linewidth=2)
plt.title('带标记的折线图')
plt.grid(True, alpha=0.3)
plt.show()
```

## 2. Matplotlib 绘图线

`linestyle` 设置线型，`linewidth` 设置线宽，`color` 设置颜色。常见线型有实线 `-`、虚线 `--`、点划线 `-.` 和点线 `:`。

```python
x = np.arange(0, 6)
plt.plot(x, x + 1, color='steelblue', linestyle='-', linewidth=2, label='实线')
plt.plot(x, x + 2, color='tomato', linestyle='--', linewidth=2, label='虚线')
plt.plot(x, x + 3, color='seagreen', linestyle=':', linewidth=3, label='点线')
plt.legend()
plt.title('不同线型')
plt.show()
```

## 3. Matplotlib 轴标签和标题

使用 `xlabel()`、`ylabel()` 设置坐标轴标签，使用 `title()` 设置标题。`loc` 可以控制标题的位置。

```python
months = ['一月', '二月', '三月', '四月']
sales = [120, 150, 135, 180]
plt.plot(months, sales, marker='o')
plt.xlabel('月份')
plt.ylabel('销售额（万元）')
plt.title('季度销售趋势', loc='center')
plt.show()
```

## 4. Matplotlib 网格线

`grid()` 可以显示网格线，`axis='x'` 或 `axis='y'` 可只显示一个方向的网格，也可以设置颜色、线型和透明度。

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title('带网格线的正弦曲线')
plt.grid(axis='both', color='gray', linestyle='--', alpha=0.5)
plt.show()
```

## 5. Matplotlib 绘制多图

`subplot(rows, cols, index)` 可以把画布划分为多个区域；更推荐使用 `subplots()` 一次创建多个坐标轴对象。

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
x = np.linspace(0, 2 * np.pi, 200)
axes[0].plot(x, np.sin(x), color='royalblue')
axes[0].set_title('sin(x)')
axes[1].plot(x, np.cos(x), color='darkorange')
axes[1].set_title('cos(x)')
for ax in axes:
    ax.grid(True, alpha=0.3)
fig.suptitle('两个子图')
plt.tight_layout()
plt.show()
```

## 6. Matplotlib 散点图

`scatter()` 用于绘制散点图。`s` 控制点的大小，`c` 控制颜色，`alpha` 控制透明度。

```python
rng = np.random.default_rng(42)
x = rng.normal(50, 10, 60)
y = 0.8 * x + rng.normal(0, 8, 60)
colors = rng.uniform(0, 1, 60)
plt.scatter(x, y, s=60, c=colors, cmap='viridis', alpha=0.8)
plt.colorbar(label='颜色值')
plt.xlabel('特征 X')
plt.ylabel('特征 Y')
plt.title('散点图示例')
plt.show()
```

## 7. Matplotlib 柱形图

`bar()` 绘制垂直柱形图，`barh()` 绘制水平柱形图。柱子的高度由数据值决定。

```python
subjects = ['数学', '语文', '英语', '科学']
scores = [92, 88, 95, 86]
plt.bar(subjects, scores, color=['#4C78A8', '#F58518', '#54A24B', '#E45756'])
plt.ylim(0, 100)
plt.ylabel('分数')
plt.title('各科成绩')
plt.show()
```

## 8. Matplotlib 饼图

`pie()` 用于展示各类别占总体的比例。`labels` 设置标签，`autopct` 显示百分比，`explode` 可以突出某一块。

```python
labels = ['Python', 'JavaScript', 'Java', '其他']
sizes = [40, 25, 20, 15]
explode = [0.06, 0, 0, 0]
plt.pie(sizes, labels=labels, explode=explode, autopct='%1.1f%%',
        startangle=90, shadow=False)
plt.title('编程语言偏好')
plt.axis('equal')
plt.show()
```

## 9. Matplotlib 直方图

`hist()` 将连续数据划分到若干区间（bins），统计每个区间中的数据数量，适合观察数据分布。

```python
rng = np.random.default_rng(7)
data = rng.normal(loc=70, scale=10, size=1000)
plt.hist(data, bins=20, color='mediumpurple', edgecolor='white', alpha=0.85)
plt.xlabel('数值')
plt.ylabel('频数')
plt.title('正态分布样本的直方图')
plt.show()
```

## 10. Matplotlib `imshow()`

`imshow()` 可以把二维数组显示为图像，常用于热力图和矩阵可视化。`cmap` 设置颜色映射，`colorbar()` 显示数值与颜色的对应关系。

```python
matrix = np.arange(100).reshape(10, 10)
plt.imshow(matrix, cmap='viridis', aspect='auto')
plt.colorbar(label='矩阵值')
plt.title('二维数组可视化')
plt.xlabel('列')
plt.ylabel('行')
plt.show()
```
